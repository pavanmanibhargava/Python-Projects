```python
Name=input("Name:")

def get_marks(subject):
    while True:
        try:
            marks=int(input(f"{subject} marks:"))
            if marks>=0 and marks<=100:
              return marks
            print("Marks must be between 0 to 100")
        
        except ValueError:
              print("Enter Valid Number")
    
    
Maths=get_marks("Maths")
Chemistry=get_marks("Chemistry")
Physics=get_marks("Physics")


Total=Maths+Chemistry+Physics
Average=(Total/3)

print("Total:",Total)
print("Average:",Average)

if Average>=90:
    print("Grade:A")
elif Average>=80:
    print("Grade:B")
elif Average>=70:
    print("Grade:C")
elif Average>=60:
    print("Grade:D")
elif Average>=50:
    print("Grade:E")
else:
    print("Grade:F")
```

```python
print("----- Expense & Budget Analyzer -----")

def get_positive_number(number):
    while True:
        try:
            positive_number=int(input(number))
            if positive_number>0:
                return positive_number
            else:
                print("Enter Positive Number")
        except ValueError:
            print("Enter Valid Number")

budget=get_positive_number("Enter Total Budget:")
no_of_expenses=get_positive_number("Enter Number of expenses:")

def get_expenses(prompt):
    expenses={}
    for i in range(prompt):
        name=input("Enter Expense Name:")
        while True:
            try:
                amount=int(input("Enter Amount:"))
                if amount>0:
                    break
                else:
                    print("Enter Positive Number")
            except ValueError:
                print("Enter Valid Number")
        if name in expenses:
            expenses[name]+=amount
        else:
            expenses[name]=amount
    return expenses

expenses=get_expenses(no_of_expenses)

def calculate_summary(expenses, budget):
    total_expenses = sum(expenses.values())
    remaining_budget = budget - total_expenses
    budget_used = round((total_expenses / budget) * 100, 2)
    average_expense = round(total_expenses / len(expenses), 2)
    categories = len(expenses)

    return (
        total_expenses,
        remaining_budget,
        budget_used,
        average_expense,
        categories
    )

#total_expenses,remaining_budget,budget_used,average_expense,categories=calculate_summary(expenses,budget) One way to store multiple returned values

calculations=calculate_summary(expenses,budget)

def display_summary(expenses,budget,calculations):
    print("----- Budget Summary -----")
    for key,value in expenses.items():
        print(f"{key}:{value}")
    total_expenses,remaining_budget,budget_used,average_expense,categories = calculations
    print(f"Total Expenses:{total_expenses}")
    print(f"Remaining Budget:{remaining_budget}")
    print(f"Budget used:{budget_used}")
    print(f"Average Expense:{average_expense}")
    print(f"Categories:{categories}")
    if total_expenses<=budget:
            print("Status: Within Budget")
    else:
            print("Status: Over Budget")


display_summary(expenses,budget,calculations)


def get_highest(expenses):
    max_key=next(iter(expenses))
    max_value=expenses[max_key]

    for key,value in expenses.items():
      if value>max_value:
           max_key=key
           max_value=value
    return f"Highest Expense: {max_key}:{max_value}"

def get_lowest(expenses):
    min_key=next(iter(expenses))
    min_value=expenses[min_key]

    for key,value in expenses.items():
      if value<min_value:
           min_key=key
           min_value=value
    return f"Lowest Expense: {min_key}:{min_value}"

highest_expense=get_highest(expenses)
lowest_expense=get_lowest(expenses)

print(highest_expense)
print(lowest_expense)

def sort_expenses(expenses):
    items=list(expenses.items())
    sorted_items=sorted(items,key=lambda item:item[1],reverse=True)
    for name, amount in sorted_items:
      print(f"{name}: {amount}")



sort_expenses(expenses)


def min_amount(expenses):
    expense={}
    min_amount=int(input("Enter Minimum amount:"))
    for key,value in expenses.items():
        if value>min_amount:
            expense[key]=value
    return expense



minimum_amount=min_amount(expenses)
print(minimum_amount)

    
def category_percentages(expenses):
    result=""
    total=sum(expenses.values())
    for key,value in expenses.items():
        percentage=round((value/total)*100,2)
        result+=f"{key}:{percentage}%\n"
    return result


filtering=category_percentages(expenses)
print(filtering)
```
