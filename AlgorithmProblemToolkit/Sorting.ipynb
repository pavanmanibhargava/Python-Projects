# 1 Bubble Sort

### 1.1 Basic Bubble Sort Implementation
Sorts the list by repeatedly comparing adjacent elements and swapping them when they are in the wrong order.

```python
numbers = [7, 2, 9, 1, 5]

for i in range(len(numbers)-1):
    for j in range(len(numbers)-1-i):
        if numbers[j]>numbers[j+1]:
            numbers[j],numbers[j+1]=numbers[j+1],numbers[j]

print(numbers)
```

### 1.2 Optimized Bubble Sort with Early Termination

Stops the algorithm early when no swaps occur during a pass, indicating that the list is already sorted.

```python
numbers = [7, 2, 9, 1, 5]

for i in range(len(numbers)-1):
    swapped=False
    for j in range(len(numbers)-1-i):
        if numbers[j]>numbers[j+1]:
            numbers[j],numbers[j+1]=numbers[j+1],numbers[j]
            swapped=True
    if not swapped:
        break
    print(f"{i+1} Pass Swapping:{swapped}")    


```

## 2. Selection Sort

### 2.1 Basic Selection Sort Implementation

```python
arr=[64,25,12,22,11]

for i in range(len(arr)):
    min_index=i
    for j in range(i+1,len(arr)):
        if arr[j]<arr[min_index]:
            min_index=j
    arr[i],arr[min_index]=arr[min_index],arr[i]

print(arr)
```
