```python
numbers = [3, 7, 11, 15, 19]
target = int(input("Enter Target:"))



def search_insert(numbers,target):
    left=0
    right=len(numbers)-1
    while left<=right:
        mid=(left+right)//2
        if numbers[mid]<target:
            left=mid+1
        elif numbers[mid]>target:
            right=mid-1
        else:
            return mid
    return left

index=search_insert(numbers,target)
print(index)
```
