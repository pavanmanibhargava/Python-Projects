Binary Search

```python
numbers = [3, 7, 11, 15, 19, 24, 31, 42, 50]

left=0
right=len(numbers)-1
target=24
found=0

while left<=right:
  
    mid=(left+right)//2
    if numbers[mid]<target:
        left=mid+1
    elif numbers[mid]>target:
        right=mid-1
    else:
       print(mid)
       found=1
       break


if found==0:
    print(-1)
```
