```python
text=input("Enter text:")
for char in ".,:;!?":
    text=text.replace(char," ")
words=text.split()
frequency={}
for word in words:
    if word.lower() in frequency:
        frequency[word.lower()]+=1
    else:
        frequency[word.lower()]=1
print(frequency)
items=list(frequency.items())
print(f"items:{items}")
sorted_list=sorted(items,key=lambda item:item[1],reverse=True)
print(f"sorted_list:{sorted_list}")
number=int(input("How many top words do you want?"))
for key,value in sorted_list[:number]:
    print(f"{key}:{value}")
```

```python
print("===== TEXT ANALYSIS =====\\n")

text=input("Enter text:")

print(f"Characters:{len(text)}")


def clean_text(text):
    for char in ',.:;!?':
        text=text.replace(char," ")
    return text.lower()

cleaned_text=clean_text(text)


def get_words(cleaned_text):
    words=cleaned_text.split()
    return words

words=get_words(cleaned_text)
print(f"Number of Words:{len(words)}")


def unique_words(words):
    set1=set()
    for word in words:
       set1.add(word)
    return set1

unique=unique_words(words)
print(f"Unique words:{len(unique)}")


def count_sentences(text):
    sentences=text.split(".")
    return len(sentences)-1

sentences=count_sentences(text)
print(f"Sentences:{sentences}")


def count_frequency(words):
    frequency={}
    for word in words:
        if word in frequency:
            frequency[word]+=1
        else:
            frequency[word]=1
    return frequency

frequency=count_frequency(words)


def sort_by_frequency(frequency):
    items=list(frequency.items())
    sorted_items=sorted(items,key=lambda item:item[1],reverse=True)
    return sorted_items

sorted_items=sort_by_frequency(frequency)
print("Most Common Words:")
for key,value in sorted_items:
    print(f"{key}:{value}")


def filter_by_frequency(frequency,min_value):
    filtered_frequency={}
    for key,value in frequency.items():
        if value>=min_value:
            filtered_frequency[key]=value
    return filtered_frequency

min_value=int(input("Enter Minimum Value:"))
filtering=filter_by_frequency(frequency,min_value)
print(f"Words occurring at least {min_value} times:")
print(filtering)
```
